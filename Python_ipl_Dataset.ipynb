```python
#Line plot
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

a = pd.read_csv("/content/IPL_Player_Stats.csv")
sns.lineplot(x="Player", y="Runs", data=a ,color='red',label = 'Runs')

plt.title("Player vs Runs ")
plt.legend(loc = "upper right")

plt.show()
```

```python
#Scatter plot


import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

a = pd.read_csv("/content/IPL_Player_Stats.csv")
sns.scatterplot(x="Matches", y="Wickets", data=a ,color='blue',label = 'Wickets')

plt.title("Wickets taken")
plt.legend(loc = "upper left")

plt.show()
```

```python
#Bar plot--1


import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

a = pd.read_csv("/content/IPL_Player_Stats.csv")
sns.barplot(x="Player", y="Strike_Rate", data=a ,color='red',label = 'Strike rate ')

plt.title("Strike rate ")

plt.legend(loc = "upper right")

plt.show()
```

```python
#Histogram

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

a = pd.read_csv("/content/IPL_Player_Stats.csv")
sns.histplot(x="Runs",  data=a ,color='blue')

plt.title("Runs Histogram")


plt.legend(loc = "upper left")

plt.show()
```

```python
#Count plot

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

a = pd.read_csv("/content/IPL_Player_Stats.csv")
sns.countplot(x="Role", data=a ,color='red',label = 'Role')

plt.title("Roles of palyers")

plt.legend(loc = "upper right")

plt.show()
```

```python
#Heat Map--1


import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

a = pd.read_csv("/content/IPL_Player_Stats.csv")

c = a.corr(numeric_only=True)
sns.heatmap(c,annot=True , cmap ="inferno" )
plt.legend()
plt.show()
```

```python
#Heat map -- 2


import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

a = pd.read_csv("/content/IPL_Player_Stats.csv")

c = a.corr(numeric_only=True)
sns.heatmap(c,annot=False , cmap ="inferno" )
plt.legend()
plt.show()
```
